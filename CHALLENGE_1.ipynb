```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
tienda1 = pd.read_csv("/content/tienda_1 .csv")
tienda2 = pd.read_csv("/content/tienda_2.csv")
tienda3 = pd.read_csv("/content/tienda_3.csv")
tienda4 = pd.read_csv("/content/tienda_4.csv")
```

```python
tienda1.head()
```

```python
tiendas = {
    "Tienda 1": tienda1,
    "Tienda 2": tienda2,
    "Tienda 3": tienda3,
    "Tienda 4": tienda4
}
```

```python
tienda1.columns
```

```python
ingresos = {}

for nombre, df in tiendas.items():
    ingresos[nombre] = df["Precio"].sum()

plt.figure()
plt.bar(ingresos.keys(), ingresos.values())
plt.title("Ingresos por tienda")
plt.ylabel("Ingresos")
plt.show()
```

```python
envio_promedio = {}

for nombre, df in tiendas.items():
    envio_promedio[nombre] = df["Costo de envío"].mean()

plt.figure()
plt.bar(envio_promedio.keys(), envio_promedio.values())
plt.title("Costo promedio de envío por tienda")
plt.ylabel("Costo de envío")
plt.show()
```

```python
resena_promedio = {}

for nombre, df in tiendas.items():
    resena_promedio[nombre] = df["Calificación"].mean()

plt.figure()
plt.bar(resena_promedio.keys(), resena_promedio.values())
plt.title("Reseña promedio por tienda")
plt.ylabel("Calificación")
plt.show()
```

```python
for nombre, df in tiendas.items():
    print(nombre)
    print("Ingresos totales:", df["Precio"].sum())
    print("Costo promedio de envío:", df["Costo de envío"].mean())
    print("Reseña promedio:", df["Calificación"].mean())
    print("-----------------------------")
```
